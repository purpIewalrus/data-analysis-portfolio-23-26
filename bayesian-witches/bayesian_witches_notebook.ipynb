## 0. Problematic

**Can we predict the probability of an individual being accused of a demonic pact based on their socio-economic characteristics?**

We anticipate that if this is the case, we should observe: 

- **P1​:** Being female should increase the probability of accusation.
- **P2:** Being single should increase the probability of accusation.
- **P3:** Being in an irregular union should increase the probability of accusation relative to socially normative marital situations.
- **P4:** The probability of accusation should increase as age increases.

## 1. Library Import

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler

# Libraries for Bayesian modeling
import pymc as pm
import arviz as az

# Librairies for visualisations
import seaborn as sns
import matplotlib.pyplot as plt
import networkx as nx
```

## 2. Directed Acyclic Graph

```python
# Create directed graph
G = nx.DiGraph()

# Define nodes
nodes = [
    'Sex',
    'Age',
    'Marital\nstatus',
    'Socio-economic\nstatus',
    'Probability of\naccusation'
]

G.add_nodes_from(nodes)

# Define causal links
edges = [
    ('Sex', 'Probability of\naccusation'),
    ('Age', 'Probability of\naccusation'),
    ('Marital\nstatus', 'Probability of\naccusation'),
    ('Socio-economic\nstatus', 'Probability of\naccusation'),
    ('Socio-economic\nstatus', 'Marital\nstatus')
]

G.add_edges_from(edges)

# Layout
pos = nx.spring_layout(G, seed=42)

# Draw graph
plt.figure(figsize=(8,6))

nx.draw(
    G,
    pos,
    with_labels=True,
    node_size=5000,
    node_color='lightblue',
    arrows=True,
    arrowsize=20,
    font_size=10,
    edge_color='gray'
)

plt.title("Directed acyclic graph representing the hypothesised\n"
    "causal relationships between socio-economic characteristics\n"
    "and the probability of being accused of a demonic pact.", fontsize=12)

plt.tight_layout()
plt.show()
```

## 6. Bayesian Logistic Regression Model

To test our hypotheses, we performed a Bayesian logistic regression using the PyMC library. Within this probabilistic framework, the posterior probability distribution of each regression parameter, conditioned on the observed data, is estimated using a Markov Chain Monte Carlo (MCMC) sampling algorithm.

The model is defined as follows.

- **Likelihood:**

$$
y_i \sim \mathrm{Bernoulli}(p_i)
$$

- **Link function:**

$$
\mathrm{logit}(p_i) = \alpha + \sum_{j=1}^{K} \beta_j X_{i,j}
$$

- **Priors:**

$$
\alpha \sim \mathcal{N}(0,1)
$$

$$
\beta_j \sim \mathcal{N}(0,1),
\qquad \forall j \in \{1,\ldots,K\}
$$

where:

- $y_i \in \{0,1\}$ is the observed outcome for individual $i$ (1 if the individual was accused of a demonic pact, 0 otherwise);
- $p_i$ is the probability that individual $i$ was accused of a demonic pact;
- $\alpha$ is the intercept of the logistic regression model;
- $\beta_j$ is the regression coefficient associated with predictor $j$;
- $X_{i,j}$ is the value of predictor $j$ for individual $i$;
- $K$ is the total number of predictors included in the model.

In our analysis, the predictors correspond to the socio-economic characteristics available in the dataset, namely age, sex, and marital status (including the dummy variables generated for categorical features).

```python
# Data standardization
scaler = StandardScaler()
X = scaler.fit_transform(X)
```

```python
n_features = X.shape[1]

with pm.Model() as model:
    
    # Priors
    beta = pm.Normal("beta", mu=0, sigma=1, shape=n_features)
    intercept = pm.Normal("intercept", mu=0, sigma=1)
    
    # Logit
    logit_p = intercept + pm.math.dot(X, beta)
    
    # Likelihood
    y_obs = pm.Bernoulli("y_obs", logit_p=logit_p, observed=y)
    
    # Sampling
    trace = pm.sample(
        draws=500,
        tune=500,
        chains=2,
        cores=1,
        target_accept=0.9
    )
```

```python
feature_names = df_encoded[variables].columns.tolist()

idata = trace

# Identify the dimension associated with the regression coefficients
beta_dims = idata.posterior["beta"].dims
dim_name = idata.posterior["beta"].dims[-1]

idata.posterior = idata.posterior.assign_coords(
    {dim_name: feature_names}
)

# Plot posterior distributions of all regression coefficients
az.plot_posterior(idata, var_names=["beta", "intercept"])
plt.tight_layout()
```

```python
az.summary(idata)
```

## 7. Results

```python
summary = az.summary(idata)

beta_age = summary.loc["beta[Age]", "mean"]
beta_single = summary.loc["beta[MaritalStatus_Single]", "mean"]
beta_irregular = summary.loc["beta[MaritalStatus_Irregular union]", "mean"]
beta_male = summary.loc["beta[Sex_Male]", "mean"]

# Hypotheses evaluation

# P1: being female increases accusations
P1_supported = beta_male < 0

# P2: being single increases accusations
P2_supported = beta_single > 0

# P3: irregular union increases accusations
P3_supported = beta_irregular > 0

# P4: being older increases accusations
P4_supported = beta_age > 0

print("P1 (being female increases accusations):", P1_supported)
print("P2 (being single increases accusations):", P2_supported)
print("P3 (irregular union increases accusations):", P3_supported)
print("P4 (being older increases accusations):", P4_supported)
```

The posterior estimates obtained from the Bayesian logistic regression suggest that most socio-economic variables have relatively uncertain effects on the probability of being accused of a demonic pact.

Regarding sex, the posterior mean associated with the variable *Sex_Male* is very close to zero (β = -0.006), and its 94% HDI largely overlaps zero. This indicates that the observed data do not provide substantial posterior evidence for a meaningful difference in accusation probability between men and women. Consequently, hypothesis **P1**, stating that women were more likely to be accused, is not strongly supported by the model.

Age appears to have a slightly negative posterior effect (β = -0.247), suggesting that older individuals may have been somewhat less likely to be accused. However, the corresponding HDI still includes zero, which indicates substantial posterior uncertainty. Therefore, the data only weakly support the existence of an age effect and does not support hypothesis **P4**.

Marital status produces more interesting posterior patterns. The coefficient associated with the category *MaritalStatus_Single* is positive (β = 0.327), meaning that single individuals may have had a higher posterior probability of accusation compared to the reference category. Although the 94% HDI still overlaps zero, the posterior distribution leans toward a positive effect, providing moderate but inconclusive support for hypothesis **P2**.

The category *MaritalStatus_Irregular* union shows the strongest effect in magnitude (β = -0.758), suggesting a potentially lower probability of accusation relative to the baseline category. Nevertheless, uncertainty remains important, as the HDI still includes zero. The categories *MaritalStatus_Married* and *MaritalStatus_Widowed* both exhibit posterior distributions centered close to zero, indicating little evidence for a substantial influence on accusation probability (**P3**).

Overall, the posterior distributions suggest that while some socio-economic characteristics may have influenced accusations, the available data do not provide strong posterior certainty for most effects included in the model.
